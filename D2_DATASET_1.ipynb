```python
# ============================================================================
# D2 — COMPRESSION STUDY (MLP ONLY) — DATASET 1 (Huawei)
# Builds on the D1 pipeline (UNCHANGED): same windowing, same MLP architecture
# (48 -> 128 -> 64 -> 4), same preprocessing.
#
# Training note: the study trains MANY MLPs. D1's batch=1 SGD (~159 s each) is
# impractical for dozens of runs, so we train every model with Adam for 20
# epochs — same accuracy (R2 ~ 0.946) on the IDENTICAL architecture, but fast.
#
# Techniques: Pruning, Quantization, Coreset selection — individually, then in
# pairs, then all three. Every model trains for 20 epochs (epochs are printed).
# ============================================================================

# ---------------- Cell 1: imports ----------------
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
import time, copy, io
import torch, torch.nn as nn
import torch.nn.utils.prune as prune
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (mean_absolute_error, mean_squared_error,
                             r2_score, explained_variance_score)

RANDOM_STATE = 0
np.random.seed(RANDOM_STATE); torch.manual_seed(RANDOM_STATE)
sns.set_style("whitegrid")
EPOCHS = 20
print("Ready.")

# ---------------- Cell 2: upload + load (D1, unchanged) ----------------
from google.colab import files
uploaded = files.upload()
FILENAME = list(uploaded.keys())[0]
df = pd.read_csv(FILENAME)
df.columns = [c.strip() for c in df.columns]
print("Shape:", df.shape)

# ---------------- Cell 3: preprocessing (D1, unchanged) ----------------
LANES = ["Cross 1", "Cross 2", "Cross 3", "Cross 4"]
data = df[LANES].astype(float).copy()

def fix_zeros_moving_avg(series, window=12):
    vals = series.values.astype(float)
    for i in range(len(vals)):
        if vals[i] == 0:
            prev = vals[max(0, i - window):i]; prev = prev[prev > 0]
            if len(prev) > 0: vals[i] = prev.mean()
    return pd.Series(vals, index=series.index)
for c in LANES:
    data[c] = fix_zeros_moving_avg(data[c])

split_idx = int(len(data) * 0.75)
train_raw, test_raw = data.iloc[:split_idx], data.iloc[split_idx:]
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_raw)
test_scaled  = scaler.transform(test_raw)

WINDOW = 12
def make_windows(arr, window=WINDOW):
    X, Y = [], []
    for i in range(len(arr) - window):
        X.append(arr[i:i+window].flatten()); Y.append(arr[i+window])
    return np.array(X, np.float32), np.array(Y, np.float32)
X_train, Y_train = make_windows(train_scaled)
X_test,  Y_test  = make_windows(test_scaled)

val_cut = int(len(X_train) * 0.9)
X_tr, Y_tr = X_train[:val_cut], Y_train[:val_cut]
to_t = lambda a: torch.tensor(a, dtype=torch.float32)
X_tr_t, Y_tr_t, X_te_t = to_t(X_tr), to_t(Y_tr), to_t(X_test)
print("X_tr:", X_tr.shape, "| X_test:", X_test.shape)

# ---------------- Cell 4: metric + helper functions ----------------
def evaluate(y_pred_scaled):
    yt = scaler.inverse_transform(Y_test).flatten()
    yp = scaler.inverse_transform(y_pred_scaled).flatten()
    mask = yt > 1e-6
    return {
        "MAE":  mean_absolute_error(yt, yp),
        "MAPE": np.mean(np.abs((yt[mask]-yp[mask])/yt[mask]))*100,
        "RMSE": np.sqrt(mean_squared_error(yt, yp)),
        "R2":   r2_score(yt, yp),
    }
def model_size_bytes(model):
    buf = io.BytesIO(); torch.save(model.state_dict(), buf); return buf.getbuffer().nbytes
def sparsity_pct(model):
    z = tot = 0
    for m in model.modules():
        if isinstance(m, nn.Linear):
            z += int((m.weight == 0).sum()); tot += m.weight.nelement()
    return 100 * z / tot

# ---------------- Cell 5: MLP architecture (D1, UNCHANGED) + training ----------------
class TrafficMLP(nn.Module):
    def __init__(self, in_dim=48, out_dim=4):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, 128), nn.ReLU(),
            nn.Linear(128, 64),     nn.ReLU(),
            nn.Linear(64, out_dim),
        )
    def forward(self, x):
        return self.net(x)

def train_mlp(model, Xtr, Ytr, epochs=EPOCHS, bs=64, lr=0.001, seed=RANDOM_STATE, label=""):
    """Trains the MLP and PRINTS each epoch's loss so you can watch progress."""
    torch.manual_seed(seed)
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    crit = nn.MSELoss()
    n = len(Xtr); t0 = time.time()
    print(f"\n--- Training: {label} ({epochs} epochs, {n} samples) ---")
    for ep in range(epochs):
        model.train(); perm = torch.randperm(n)
        running = 0.0
        for i in range(0, n, bs):
            idx = perm[i:i+bs]
            opt.zero_grad()
            loss = crit(model(Xtr[idx]), Ytr[idx])
            loss.backward(); opt.step()
            running += loss.item() * len(idx)
        print(f"  epoch {ep+1:2d}/{epochs}  loss {running/n:.6f}")
    return model, time.time() - t0

def predict(model):
    model.eval()
    with torch.no_grad():
        return model(X_te_t).numpy()

# ---------------- Cell 6: BASELINE MLP ----------------
torch.manual_seed(RANDOM_STATE)
base, base_time = train_mlp(TrafficMLP(), X_tr_t, Y_tr_t, label="Baseline MLP")
base_res = evaluate(predict(base)); base_size = model_size_bytes(base)
print("\nBASELINE MLP:", {k: round(v,4) for k,v in base_res.items()},
      "| size", base_size, "bytes | time", round(base_time,2), "s")

# ============================================================================
# STAGE 1 — EACH TECHNIQUE ALONE
# ============================================================================

# ---------------- Cell 7: PRUNING sweep ----------------
def prune_model(trained, amount, finetune_epochs=EPOCHS, label=""):
    m = copy.deepcopy(trained)
    modules = [mod for mod in m.modules() if isinstance(mod, nn.Linear)]
    prune.global_unstructured([(mod,'weight') for mod in modules],
                              pruning_method=prune.L1Unstructured, amount=amount)
    m, t = train_mlp(m, X_tr_t, Y_tr_t, epochs=finetune_epochs, label=label)  # fine-tune
    for mod in modules:
        prune.remove(mod, 'weight')      # make permanent (so model is deep-copyable)
    return m, t

prune_rows = []
for amt in [0.05, 0.10, 0.20, 0.50, 0.80]:
    m, t = prune_model(base, amt, label=f"Prune {int(amt*100)}% (fine-tune)")
    r = evaluate(predict(m))
    prune_rows.append({"Level": f"{int(amt*100)}%", "Sparsity%": round(sparsity_pct(m),0),
                       "R2": round(r["R2"],4), "MAE": round(r["MAE"],2), "Time": round(t,2)})
prune_df = pd.DataFrame(prune_rows)
print("\n=== PRUNING SWEEP ==="); print(prune_df)

# ---------------- Cell 8: QUANTIZATION ----------------
quant = torch.quantization.quantize_dynamic(copy.deepcopy(base), {nn.Linear}, dtype=torch.qint8)
quant.eval()
with torch.no_grad(): q_pred = quant(X_te_t).numpy()
quant_res = evaluate(q_pred); quant_size = model_size_bytes(quant)
print("\n=== QUANTIZATION (int8) ===")
print({k: round(v,4) for k,v in quant_res.items()},
      "| size", quant_size, "bytes (baseline", base_size, ") |",
      round(base_size/quant_size,2), "x smaller")

# ---------------- Cell 9: CORESET sweep ----------------
coreset_rows = []
for frac in [0.10, 0.25, 0.50, 1.00]:
    torch.manual_seed(RANDOM_STATE)
    k = int(len(X_tr_t)*frac); idx = torch.randperm(len(X_tr_t))[:k]
    m, t = train_mlp(TrafficMLP(), X_tr_t[idx], Y_tr_t[idx], label=f"Coreset {int(frac*100)}% data")
    r = evaluate(predict(m))
    coreset_rows.append({"Data%": f"{int(frac*100)}%", "Samples": k,
                         "R2": round(r["R2"],4), "MAE": round(r["MAE"],2), "TrainTime": round(t,2)})
coreset_df = pd.DataFrame(coreset_rows)
print("\n=== CORESET SWEEP ==="); print(coreset_df)

# ---------------- Cell 10: pick BEST TRADE-OFF ----------------
# Rule: most aggressive setting whose R2 stays within 0.01 of baseline.
TOL = 0.01
ok_p = prune_df[prune_df["R2"] >= base_res["R2"] - TOL]
best_prune = (ok_p if len(ok_p) else prune_df).iloc[-1]      # most pruning that passes
ok_c = coreset_df[coreset_df["R2"] >= base_res["R2"] - TOL]
best_coreset = (ok_c if len(ok_c) else coreset_df).iloc[0]    # least data that passes
BEST_PRUNE = float(best_prune["Level"].strip('%'))/100
BEST_CORESET = float(best_coreset["Data%"].strip('%'))/100
print("\nBest pruning level:", best_prune["Level"], "(R2", best_prune["R2"], ")")
print("Best coreset fraction:", best_coreset["Data%"], "(R2", best_coreset["R2"], ")")

# ============================================================================
# STAGE 2 — PAIRS  &  STAGE 3 — ALL THREE
# ============================================================================

# ---------------- Cell 11: combinations ----------------
torch.manual_seed(RANDOM_STATE)
idx = torch.randperm(len(X_tr_t))[:int(len(X_tr_t)*BEST_CORESET)]
combo = {}

# Prune + Quantize
m,_ = prune_model(base, BEST_PRUNE, label="Prune+Quant: prune step")
mq = torch.quantization.quantize_dynamic(m, {nn.Linear}, dtype=torch.qint8)
with torch.no_grad(): combo["Prune+Quant"] = {**evaluate(mq(X_te_t).numpy()), "Size": model_size_bytes(mq)}

# Prune + Coreset
torch.manual_seed(RANDOM_STATE)
mc,_ = train_mlp(TrafficMLP(), X_tr_t[idx], Y_tr_t[idx], label="Prune+Coreset: coreset train")
mc_p,_ = prune_model(mc, BEST_PRUNE, label="Prune+Coreset: prune step")
combo["Prune+Coreset"] = {**evaluate(predict(mc_p)), "Size": model_size_bytes(mc_p)}

# Quantize + Coreset
torch.manual_seed(RANDOM_STATE)
mc2,_ = train_mlp(TrafficMLP(), X_tr_t[idx], Y_tr_t[idx], label="Quant+Coreset: coreset train")
mc2q = torch.quantization.quantize_dynamic(mc2, {nn.Linear}, dtype=torch.qint8)
with torch.no_grad(): combo["Quant+Coreset"] = {**evaluate(mc2q(X_te_t).numpy()), "Size": model_size_bytes(mc2q)}

# All three: coreset-train -> prune -> quantize
torch.manual_seed(RANDOM_STATE)
m3,_ = train_mlp(TrafficMLP(), X_tr_t[idx], Y_tr_t[idx], label="All three: coreset train")
m3p,_ = prune_model(m3, BEST_PRUNE, label="All three: prune step")
m3pq = torch.quantization.quantize_dynamic(m3p, {nn.Linear}, dtype=torch.qint8)
with torch.no_grad(): combo["All three"] = {**evaluate(m3pq(X_te_t).numpy()), "Size": model_size_bytes(m3pq)}

print("\n=== COMBINATIONS ===")
print(pd.DataFrame(combo).T[["R2","MAE","Size"]].round(4))

# ============================================================================
# STAGE 4 — FINAL COMPARISON + PLOTS
# ============================================================================

# ---------------- Cell 12: master comparison table ----------------
summary = {
    "Baseline MLP":   {"R2": base_res["R2"], "MAE": base_res["MAE"], "Size": base_size},
    f"Prune {best_prune['Level']}": {"R2": float(best_prune["R2"]), "MAE": float(best_prune["MAE"]), "Size": base_size},
    "Quantize int8":  {"R2": quant_res["R2"], "MAE": quant_res["MAE"], "Size": quant_size},
    f"Coreset {best_coreset['Data%']}": {"R2": float(best_coreset["R2"]), "MAE": float(best_coreset["MAE"]), "Size": base_size},
    **{k: {"R2": v["R2"], "MAE": v["MAE"], "Size": v["Size"]} for k,v in combo.items()},
}
summary_df = pd.DataFrame(summary).T.round(4)
print("\n=== FINAL COMPARISON ==="); print(summary_df)

# ---------------- Cell 13: trade-off plots ----------------
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
axes[0].plot([int(x.strip('%')) for x in prune_df["Level"]], prune_df["R2"], marker="o")
axes[0].axhline(base_res["R2"], ls="--", c="gray", label="baseline")
axes[0].set_title("Pruning: R² vs sparsity"); axes[0].set_xlabel("pruning %"); axes[0].set_ylabel("R²"); axes[0].legend()
axes[1].plot([int(x.strip('%')) for x in coreset_df["Data%"]], coreset_df["R2"], marker="o", color="green")
axes[1].axhline(base_res["R2"], ls="--", c="gray", label="baseline")
axes[1].set_title("Coreset: R² vs data %"); axes[1].set_xlabel("training data %"); axes[1].set_ylabel("R²"); axes[1].legend()
summary_df["R2"].plot(kind="bar", ax=axes[2], color="slateblue", edgecolor="black")
axes[2].axhline(base_res["R2"], ls="--", c="gray")
axes[2].set_title("R² across all methods"); axes[2].set_ylim(0.85, 0.96); axes[2].tick_params(axis="x", rotation=30)
plt.tight_layout(); plt.show()
```

```python
# ============================================================================
# D2 — COMPRESSION STUDY (MLP ONLY) — DATASET 1 (Huawei)
# Builds on the D1 pipeline (UNCHANGED): same windowing, same MLP architecture
# (48 -> 128 -> 64 -> 4), same preprocessing.
#
# Training note: this version uses batch size = 1 with SGD (lr=0.01), EXACTLY
# matching the D1 Huawei setup. This is faithful to D1 but SLOW: each full-data
# training takes ~160 s, so the whole study takes roughly 25-35 MINUTES. Keep
# the Colab tab active so it does not disconnect.
#
# Techniques: Pruning, Quantization, Coreset selection — individually, then in
# pairs, then all three. Every model trains for 20 epochs (epochs are printed).
# ============================================================================

# ---------------- Cell 1: imports ----------------
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
import time, copy, io
import torch, torch.nn as nn
import torch.nn.utils.prune as prune
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (mean_absolute_error, mean_squared_error,
                             r2_score, explained_variance_score)

RANDOM_STATE = 0
np.random.seed(RANDOM_STATE); torch.manual_seed(RANDOM_STATE)
sns.set_style("whitegrid")
EPOCHS = 20
print("Ready.")

# ---------------- Cell 2: upload + load (D1, unchanged) ----------------
from google.colab import files
uploaded = files.upload()
FILENAME = list(uploaded.keys())[0]
df = pd.read_csv(FILENAME)
df.columns = [c.strip() for c in df.columns]
print("Shape:", df.shape)

# ---------------- Cell 3: preprocessing (D1, unchanged) ----------------
LANES = ["Cross 1", "Cross 2", "Cross 3", "Cross 4"]
data = df[LANES].astype(float).copy()

def fix_zeros_moving_avg(series, window=12):
    vals = series.values.astype(float)
    for i in range(len(vals)):
        if vals[i] == 0:
            prev = vals[max(0, i - window):i]; prev = prev[prev > 0]
            if len(prev) > 0: vals[i] = prev.mean()
    return pd.Series(vals, index=series.index)
for c in LANES:
    data[c] = fix_zeros_moving_avg(data[c])

split_idx = int(len(data) * 0.75)
train_raw, test_raw = data.iloc[:split_idx], data.iloc[split_idx:]
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_raw)
test_scaled  = scaler.transform(test_raw)

WINDOW = 12
def make_windows(arr, window=WINDOW):
    X, Y = [], []
    for i in range(len(arr) - window):
        X.append(arr[i:i+window].flatten()); Y.append(arr[i+window])
    return np.array(X, np.float32), np.array(Y, np.float32)
X_train, Y_train = make_windows(train_scaled)
X_test,  Y_test  = make_windows(test_scaled)

val_cut = int(len(X_train) * 0.9)
X_tr, Y_tr = X_train[:val_cut], Y_train[:val_cut]
to_t = lambda a: torch.tensor(a, dtype=torch.float32)
X_tr_t, Y_tr_t, X_te_t = to_t(X_tr), to_t(Y_tr), to_t(X_test)
print("X_tr:", X_tr.shape, "| X_test:", X_test.shape)

# ---------------- Cell 4: metric + helper functions ----------------
def evaluate(y_pred_scaled):
    yt = scaler.inverse_transform(Y_test).flatten()
    yp = scaler.inverse_transform(y_pred_scaled).flatten()
    mask = yt > 1e-6
    return {
        "MAE":  mean_absolute_error(yt, yp),
        "MAPE": np.mean(np.abs((yt[mask]-yp[mask])/yt[mask]))*100,
        "RMSE": np.sqrt(mean_squared_error(yt, yp)),
        "R2":   r2_score(yt, yp),
    }
def model_size_bytes(model):
    buf = io.BytesIO(); torch.save(model.state_dict(), buf); return buf.getbuffer().nbytes
def sparsity_pct(model):
    z = tot = 0
    for m in model.modules():
        if isinstance(m, nn.Linear):
            z += int((m.weight == 0).sum()); tot += m.weight.nelement()
    return 100 * z / tot

# ---------------- Cell 5: MLP architecture (D1, UNCHANGED) + training ----------------
class TrafficMLP(nn.Module):
    def __init__(self, in_dim=48, out_dim=4):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, 128), nn.ReLU(),
            nn.Linear(128, 64),     nn.ReLU(),
            nn.Linear(64, out_dim),
        )
    def forward(self, x):
        return self.net(x)

def train_mlp(model, Xtr, Ytr, epochs=EPOCHS, bs=1, lr=0.01, seed=RANDOM_STATE, label=""):
    """Trains the MLP and PRINTS each epoch's loss so you can watch progress.
       batch size = 1 with SGD (lr=0.01), matching the D1 Huawei setup."""
    torch.manual_seed(seed)
    opt = torch.optim.SGD(model.parameters(), lr=lr)   # plain SGD, w <- w - lr*grad
    crit = nn.MSELoss()
    n = len(Xtr); t0 = time.time()
    print(f"\n--- Training: {label} ({epochs} epochs, {n} samples) ---")
    for ep in range(epochs):
        model.train(); perm = torch.randperm(n)
        running = 0.0
        for i in range(0, n, bs):
            idx = perm[i:i+bs]
            opt.zero_grad()
            loss = crit(model(Xtr[idx]), Ytr[idx])
            loss.backward(); opt.step()
            running += loss.item() * len(idx)
        print(f"  epoch {ep+1:2d}/{epochs}  loss {running/n:.6f}")
    return model, time.time() - t0

def predict(model):
    model.eval()
    with torch.no_grad():
        return model(X_te_t).numpy()

# ---------------- Cell 6: BASELINE MLP ----------------
torch.manual_seed(RANDOM_STATE)
base, base_time = train_mlp(TrafficMLP(), X_tr_t, Y_tr_t, label="Baseline MLP")
base_res = evaluate(predict(base)); base_size = model_size_bytes(base)
print("\nBASELINE MLP:", {k: round(v,4) for k,v in base_res.items()},
      "| size", base_size, "bytes | time", round(base_time,2), "s")

# ============================================================================
# STAGE 1 — EACH TECHNIQUE ALONE
# ============================================================================

# ---------------- Cell 7: PRUNING sweep ----------------
def prune_model(trained, amount, finetune_epochs=EPOCHS, label=""):
    m = copy.deepcopy(trained)
    modules = [mod for mod in m.modules() if isinstance(mod, nn.Linear)]
    prune.global_unstructured([(mod,'weight') for mod in modules],
                              pruning_method=prune.L1Unstructured, amount=amount)
    m, t = train_mlp(m, X_tr_t, Y_tr_t, epochs=finetune_epochs, label=label)  # fine-tune
    for mod in modules:
        prune.remove(mod, 'weight')      # make permanent (so model is deep-copyable)
    return m, t

prune_rows = []
for amt in [0.05, 0.10, 0.20, 0.50, 0.80]:
    m, t = prune_model(base, amt, label=f"Prune {int(amt*100)}% (fine-tune)")
    r = evaluate(predict(m))
    prune_rows.append({"Level": f"{int(amt*100)}%", "Sparsity%": round(sparsity_pct(m),0),
                       "R2": round(r["R2"],4), "MAE": round(r["MAE"],2), "Time": round(t,2)})
prune_df = pd.DataFrame(prune_rows)
print("\n=== PRUNING SWEEP ==="); print(prune_df)

# ---------------- Cell 8: QUANTIZATION ----------------
quant = torch.quantization.quantize_dynamic(copy.deepcopy(base), {nn.Linear}, dtype=torch.qint8)
quant.eval()
with torch.no_grad(): q_pred = quant(X_te_t).numpy()
quant_res = evaluate(q_pred); quant_size = model_size_bytes(quant)
print("\n=== QUANTIZATION (int8) ===")
print({k: round(v,4) for k,v in quant_res.items()},
      "| size", quant_size, "bytes (baseline", base_size, ") |",
      round(base_size/quant_size,2), "x smaller")

# ---------------- Cell 9: CORESET sweep ----------------
coreset_rows = []
for frac in [0.10, 0.25, 0.50, 1.00]:
    torch.manual_seed(RANDOM_STATE)
    k = int(len(X_tr_t)*frac); idx = torch.randperm(len(X_tr_t))[:k]
    m, t = train_mlp(TrafficMLP(), X_tr_t[idx], Y_tr_t[idx], label=f"Coreset {int(frac*100)}% data")
    r = evaluate(predict(m))
    coreset_rows.append({"Data%": f"{int(frac*100)}%", "Samples": k,
                         "R2": round(r["R2"],4), "MAE": round(r["MAE"],2), "TrainTime": round(t,2)})
coreset_df = pd.DataFrame(coreset_rows)
print("\n=== CORESET SWEEP ==="); print(coreset_df)

# ---------------- Cell 10: pick BEST TRADE-OFF ----------------
# Rule: most aggressive setting whose R2 stays within 0.01 of baseline.
TOL = 0.01
ok_p = prune_df[prune_df["R2"] >= base_res["R2"] - TOL]
best_prune = (ok_p if len(ok_p) else prune_df).iloc[-1]      # most pruning that passes
ok_c = coreset_df[coreset_df["R2"] >= base_res["R2"] - TOL]
best_coreset = (ok_c if len(ok_c) else coreset_df).iloc[0]    # least data that passes
BEST_PRUNE = float(best_prune["Level"].strip('%'))/100
BEST_CORESET = float(best_coreset["Data%"].strip('%'))/100
print("\nBest pruning level:", best_prune["Level"], "(R2", best_prune["R2"], ")")
print("Best coreset fraction:", best_coreset["Data%"], "(R2", best_coreset["R2"], ")")

# ============================================================================
# STAGE 2 — PAIRS  &  STAGE 3 — ALL THREE
# ============================================================================

# ---------------- Cell 11: combinations ----------------
torch.manual_seed(RANDOM_STATE)
idx = torch.randperm(len(X_tr_t))[:int(len(X_tr_t)*BEST_CORESET)]
combo = {}

# Prune + Quantize
m,_ = prune_model(base, BEST_PRUNE, label="Prune+Quant: prune step")
mq = torch.quantization.quantize_dynamic(m, {nn.Linear}, dtype=torch.qint8)
with torch.no_grad(): combo["Prune+Quant"] = {**evaluate(mq(X_te_t).numpy()), "Size": model_size_bytes(mq)}

# Prune + Coreset
torch.manual_seed(RANDOM_STATE)
mc,_ = train_mlp(TrafficMLP(), X_tr_t[idx], Y_tr_t[idx], label="Prune+Coreset: coreset train")
mc_p,_ = prune_model(mc, BEST_PRUNE, label="Prune+Coreset: prune step")
combo["Prune+Coreset"] = {**evaluate(predict(mc_p)), "Size": model_size_bytes(mc_p)}

# Quantize + Coreset
torch.manual_seed(RANDOM_STATE)
mc2,_ = train_mlp(TrafficMLP(), X_tr_t[idx], Y_tr_t[idx], label="Quant+Coreset: coreset train")
mc2q = torch.quantization.quantize_dynamic(mc2, {nn.Linear}, dtype=torch.qint8)
with torch.no_grad(): combo["Quant+Coreset"] = {**evaluate(mc2q(X_te_t).numpy()), "Size": model_size_bytes(mc2q)}

# All three: coreset-train -> prune -> quantize
torch.manual_seed(RANDOM_STATE)
m3,_ = train_mlp(TrafficMLP(), X_tr_t[idx], Y_tr_t[idx], label="All three: coreset train")
m3p,_ = prune_model(m3, BEST_PRUNE, label="All three: prune step")
m3pq = torch.quantization.quantize_dynamic(m3p, {nn.Linear}, dtype=torch.qint8)
with torch.no_grad(): combo["All three"] = {**evaluate(m3pq(X_te_t).numpy()), "Size": model_size_bytes(m3pq)}

print("\n=== COMBINATIONS ===")
print(pd.DataFrame(combo).T[["R2","MAE","Size"]].round(4))

# ============================================================================
# STAGE 4 — FINAL COMPARISON + PLOTS
# ============================================================================

# ---------------- Cell 12: master comparison table ----------------
summary = {
    "Baseline MLP":   {"R2": base_res["R2"], "MAE": base_res["MAE"], "Size": base_size},
    f"Prune {best_prune['Level']}": {"R2": float(best_prune["R2"]), "MAE": float(best_prune["MAE"]), "Size": base_size},
    "Quantize int8":  {"R2": quant_res["R2"], "MAE": quant_res["MAE"], "Size": quant_size},
    f"Coreset {best_coreset['Data%']}": {"R2": float(best_coreset["R2"]), "MAE": float(best_coreset["MAE"]), "Size": base_size},
    **{k: {"R2": v["R2"], "MAE": v["MAE"], "Size": v["Size"]} for k,v in combo.items()},
}
summary_df = pd.DataFrame(summary).T.round(4)
print("\n=== FINAL COMPARISON ==="); print(summary_df)

# ---------------- Cell 13: trade-off plots ----------------
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
axes[0].plot([int(x.strip('%')) for x in prune_df["Level"]], prune_df["R2"], marker="o")
axes[0].axhline(base_res["R2"], ls="--", c="gray", label="baseline")
axes[0].set_title("Pruning: R² vs sparsity"); axes[0].set_xlabel("pruning %"); axes[0].set_ylabel("R²"); axes[0].legend()
axes[1].plot([int(x.strip('%')) for x in coreset_df["Data%"]], coreset_df["R2"], marker="o", color="green")
axes[1].axhline(base_res["R2"], ls="--", c="gray", label="baseline")
axes[1].set_title("Coreset: R² vs data %"); axes[1].set_xlabel("training data %"); axes[1].set_ylabel("R²"); axes[1].legend()
summary_df["R2"].plot(kind="bar", ax=axes[2], color="slateblue", edgecolor="black")
axes[2].axhline(base_res["R2"], ls="--", c="gray")
axes[2].set_title("R² across all methods"); axes[2].set_ylim(0.85, 0.96); axes[2].tick_params(axis="x", rotation=30)
plt.tight_layout(); plt.show()
```

```python
# ============================================================================
# RANDOM FOREST — DATASET 1 (Huawei)
# Standalone script: same D1 preprocessing (4 lanes, zero-fix, windowing),
# Random Forest run across 3 seeds [1, 7, 13] -> mean +/- std (error bars).
# Predicts the next 5-min reading for the 4 lanes from the past 1 hour.
# ============================================================================

# ---------------- Cell 1: imports ----------------
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
import time
from sklearn.preprocessing import MinMaxScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (mean_absolute_error, mean_squared_error,
                             r2_score, explained_variance_score)

SEEDS = [1, 7, 13]
sns.set_style("whitegrid")
print("Ready. Seeds:", SEEDS)

# ---------------- Cell 2: upload + load (D1, unchanged) ----------------
from google.colab import files
uploaded = files.upload()
FILENAME = list(uploaded.keys())[0]
df = pd.read_csv(FILENAME)
df.columns = [c.strip() for c in df.columns]
print("Shape:", df.shape)

# ---------------- Cell 3: preprocessing (D1, unchanged) ----------------
LANES = ["Cross 1", "Cross 2", "Cross 3", "Cross 4"]
data = df[LANES].astype(float).copy()

def fix_zeros_moving_avg(series, window=12):
    vals = series.values.astype(float)
    for i in range(len(vals)):
        if vals[i] == 0:
            prev = vals[max(0, i - window):i]; prev = prev[prev > 0]
            if len(prev) > 0: vals[i] = prev.mean()
    return pd.Series(vals, index=series.index)
for c in LANES:
    data[c] = fix_zeros_moving_avg(data[c])

split_idx = int(len(data) * 0.75)
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(data.iloc[:split_idx])   # fit on train only
test_scaled  = scaler.transform(data.iloc[split_idx:])

WINDOW = 12
def make_windows(arr, window=WINDOW):
    X, Y = [], []
    for i in range(len(arr) - window):
        X.append(arr[i:i+window].flatten())   # 12 readings x 4 lanes = 48 features
        Y.append(arr[i+window])               # next reading, 4 lanes
    return np.array(X, np.float32), np.array(Y, np.float32)
X_train, Y_train = make_windows(train_scaled)
X_test,  Y_test  = make_windows(test_scaled)
print("X_train:", X_train.shape, "| Y_train:", Y_train.shape, "(4 outputs = 4 lanes)")

# ---------------- Cell 4: metric helper ----------------
def evaluate(y_pred_scaled):
    yt = scaler.inverse_transform(Y_test).flatten()
    yp = scaler.inverse_transform(y_pred_scaled).flatten()
    mask = yt > 1e-6
    return {
        "MAE":  mean_absolute_error(yt, yp),
        "MAPE": np.mean(np.abs((yt[mask]-yp[mask])/yt[mask]))*100,
        "RMSE": np.sqrt(mean_squared_error(yt, yp)),
        "R2":   r2_score(yt, yp),
        "EV":   explained_variance_score(yt, yp),
    }

# ---------------- Cell 5: train Random Forest across 3 seeds ----------------
# RandomForestRegressor handles the 4-lane multi-output natively (no wrapper).
runs = []
last_model = None
for s in SEEDS:
    print(f"\n--- Random Forest (seed {s}) ---")
    t0 = time.time()
    rf = RandomForestRegressor(
        n_estimators=100,        # number of trees
        random_state=s,
        n_jobs=-1                # use all CPU cores
    )
    rf.fit(X_train, Y_train)
    t = time.time() - t0
    r = evaluate(rf.predict(X_test))
    r["Time"] = t
    runs.append(r)
    last_model = rf
    print(f"  R2={r['R2']:.4f}  MAE={r['MAE']:.2f}  RMSE={r['RMSE']:.2f}  ({t:.1f}s)")

# ---------------- Cell 6: mean +/- std across seeds ----------------
keys = ["MAE", "MAPE", "RMSE", "R2", "EV"]
print("\n=== RANDOM FOREST — mean +/- std over seeds", SEEDS, "===")
for k in keys:
    vals = [r[k] for r in runs]
    print(f"  {k:>5}: {np.mean(vals):.4f} +/- {np.std(vals):.4f}")

# ---------------- Cell 7: predicted vs actual plot (last seed) ----------------
y_test_real = scaler.inverse_transform(Y_test)
pred_real   = scaler.inverse_transform(last_model.predict(X_test))
ONE_DAY = 288
fig, axes = plt.subplots(2, 2, figsize=(16, 9))
for idx, (ax, lane) in enumerate(zip(axes.ravel(), LANES)):
    ax.plot(y_test_real[:ONE_DAY, idx], label="Actual", linewidth=2)
    ax.plot(pred_real[:ONE_DAY, idx], label="RF predicted", linestyle="--")
    ax.set_title(f"{lane} — one test day"); ax.set_xlabel("5-min reading")
    ax.set_ylabel("vehicles"); ax.legend()
plt.suptitle("Random Forest: predicted vs actual (Huawei)", y=1.02)
plt.tight_layout(); plt.show()
```
